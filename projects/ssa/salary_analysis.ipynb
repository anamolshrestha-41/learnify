```python
import numpy as np
import pandas as pd
import plotly.express as px
from scipy import stats
import os
from sklearn.preprocessing import MinMaxScaler, StandardScaler
print("Libraries are successfully imported")
```

## 1. Load Dataset

```python
df= pd.read_csv("dataset.csv")
os.makedirs("./outputs", exist_ok=True)
```

## 2. Info about dataset

```python
df.shape
df.info()
df.describe()
df.isna().sum()
df.duplicated().sum()
```

## 3. Basic salary statistics

```python
mean_salary= df["salary"].mean()
print("Mean Salary: ", mean_salary)

median_salary= df["salary"].median()
print("Median Salary: ", median_salary)

mode_salary= df["salary"].mode()
print("Mode Salary: ", mode_salary)

variance_salary= df["salary"].var()
print("Salary Variance: ", variance_salary)

standard_salary= df["salary"].std()
print("Standard Salary: ", standard_salary)
```

## 4. Salary Stats

```python
salary_stats = {
    "Mean": df["salary"].mean(),
    "Median": df["salary"].median(),
    "Mode": df["salary"].mode(),
    "Variance": df["salary"].var(),
    "Std": df["salary"].std(),
}
salary_stats
```

## 4. Salary Distribution

```python
fig= px.histogram(
    df,
    x= "salary",
    nbins=5,
    title="Salary Distribution"
)

fig.show()
fig.write_image("./outputs/salary_distribution.png")
```

## 5. Salary by experience

```python
fig= px.scatter(
    df,
    x="experience",
    y="salary",
    title="Experience vs Salary"
)
fig.show()
fig.write_image("./outputs/Experience_vs_Salary.png")
correlation= df["experience"].corr(df["salary"])
print(f"Experience-Salary Correlation: ", correlation)
numeric_df = df[["age", "experience", "salary"]]
correlation_matrix = numeric_df.corr()
fig= px.imshow(
    correlation_matrix,
    title="Correlation Matrix"
)
fig.show()
fig.write_image("./outputs/correlation_matrix.png")
```

## 6. Calculate Z-score for Salary

```python
df["salary_zscore"]= stats.zscore(df["salary"])
df[["salary", "salary_zscore"]]
fig= px.bar(
    df,
    x="salary",
    y="salary_zscore",
    title="Zscore calculation of salary"
)
fig.show()
fig.write_image("./outputs/zScore_for_salary.png")
```

## 7. Outliers

```python
outliers= df[abs(df["salary_zscore"])>3]
outlier_count= (abs(df["salary_zscore"])>3).sum()
print("Nummber of Salary Outliers: ", outlier_count)
df["salary_outlier"]= abs(df["salary_zscore"])>3
df[["employee_id", "salary", "salary_zscore", "salary_outlier"]]
fig= px.box(
    df,
    y="salary",
    title="Salary Boxplot"
)
fig.show()
fig.write_image("./outputs/salary_boxplot.png")
```

## 8. Calculating covariance

```python
# cov_exp_salary= df["experience"].cov(df["salary"])
cov_exp_salary= df[["experience", "salary"]].cov()
cov_exp_salary
```

## 9. Calculate Probability

```python
total_emp= len(df)
print("Total Employees: ", total_emp)
high_salary_count=(df["salary"]>60000).sum()
print("Employees earning more than 60000: ", high_salary_count)
probability_high_salary= high_salary_count/total_emp
print(f"Probability of random selecting an employee earning more than 60000: ", probability_high_salary)

salary_range=(
    (df["salary"]>=40000) & 
    (df["salary"]<=70000)
)
range_count= salary_range.sum()
probability_range= range_count/len(df)
print( f"Probability of salary between 40000 and 70000: "
    f"{probability_range:.2%}")

experienced= df["experience"]>5
probability_experienced= experienced.mean()
print(
    f"Probability of experience > 5 years: "
    f"{probability_experienced:.2%}"
)
high_salary_given_experience=(
    (df["salary"]>60000) & experienced
).sum()
```

## 10. Calculate Conditional Proababilty

```python
experienced_count= experienced.sum()
conditional_probability=(
    high_salary_given_experience/experienced_count
)
print(
    "P(Salary > 60000 | Experience > 5):",
    conditional_probability
)

print(
    f"Percentage: {conditional_probability:.2%}"
)
```

## 11. Likelihood

```python
high_salary=df["salary"]>60000
p_experienced_given_high_salary=(
    (experienced & high_salary).sum()/high_salary.sum()
)
print(f"P(Experienced | High Salary) = {p_experienced_given_high_salary}")

p_high_salary_given_experienced=(
    p_experienced_given_high_salary*high_salary/experienced
)
print(f"P(High Salary | Experienced) = {p_high_salary_given_experienced}")
```

## 12. Confidence interval

```python
sample_mean= df["salary"].mean()
sample_std= df["salary"].std()
sample_size= df["salary"].count()

std_error= sample_std/ np.sqrt(sample_size)

confidence_interval= stats.t.interval(
    confidence=0.95,
    df= sample_size -1,
    loc= sample_mean,
    scale= std_error
)
lower_bound, upper_bound= confidence_interval
print(f" Sample Mean: {sample_mean}")
print(f"95% confidence interval: {lower_bound}, {upper_bound}")
```

## 13. Interpretation

```python
import plotly.graph_objects as go

# Display results
print(f"Mean Salary:  {sample_mean:,.2f}")
print(f"Lower Bound:  {lower_bound:,.2f}")
print(f"Upper Bound:  {upper_bound:,.2f}")

# Create figure
fig = go.Figure()

# Add mean salary with confidence interval
fig.add_trace(
    go.Scatter(
        x=["Mean Salary"],
        y=[sample_mean],
        mode="markers",
        marker=dict(size=12),
        error_y=dict(
            type="data",
            symmetric=False,
            array=[upper_bound - sample_mean],
            arrayminus=[sample_mean - lower_bound],
            visible=True,
            thickness=2,
            width=8
        ),
        hovertemplate=(
            "Mean Salary: %{y:,.2f}<br>"
            f"95% CI: {lower_bound:,.2f} – {upper_bound:,.2f}"
            "<extra></extra>"
        )
    )
)

# Customize layout
fig.update_layout(
    title="95% Confidence Interval for Mean Salary",
    xaxis_title="",
    yaxis_title="Salary",
    template="plotly_white",
    width=800,
    height=500
)

# Display
fig.show()
```

## 14. Hypothesis Testing

```python
hypothesized_mean= 60000
alpha= 0.05
t_statistic, p_value= stats.ttest_1samp(
    df["salary"],
    hypothesized_mean
)
print(f"Sample Mean: {df['salary'].mean():,.2f}")
print(f"Hypothesized Mean: {hypothesized_mean:,.2f}")
print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")
if p_value< alpha:
    print("Decision: Reject the null hypothesis")
    print(
        "There is statistically significant evidence that "
        "the mean salary differs from 60000."
    )
else:
    print("Decision: Fail to Reject the Null Hypothesis")
    print(
        "There is not enough statistical evidence to conclude "
        "that the mean salary differs from 60000."
    )
```
